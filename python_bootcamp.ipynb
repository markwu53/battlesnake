## Variables

Variables are one of the building blocks of Python. Let's define one:

````python
my_variable = "my value"
print(f"my_variable - type: {type(my_variable)} value: {my_variable}")
````

If you're familiar with other programming languages, you might be surprised that we didn't have to specify the type of
the variable. Python figured it out for us! In technical terms, Python is a "duck typed" language. If it looks like a
duck, walks like a duck, and quacks like a duck, it's a duck. This is really handy when you're prototyping and don't
quite know which types you're working with. It also leaves you open to a whole class of exceptions that strongly-typed
languages prevent. If we know what type we want to work with, we can use a "type hint":

````python
my_variable: int = "my value"
print(f"my_variable - type: {type(my_variable)} value: {my_variable}")
````

This doesn't prevent us from using the wrong variable type at runtime. The next cell will run without complaining that
we've supplied an integer instead of a string. Type hints inform our development tools about our intentions without
enforcing them. When we do something that doesn't match those intentions, our IDE will inform us. In the following cell,
your IDE should complain about the `123` value.

````python
my_variable: str = 123
print(f"my_variable - type: {type(my_variable)} value: {my_variable}")
````

Python has a variety of variable types:

````python
# "Scalar" Types
a_boolean: bool = True
a_string: str = "my string"
an_integer: int = 123
a_float: float = 3.14
a_complex_number: complex = 3.14j
a_bytes: bytes = b"my bytes"

# Container Types:
a_set: set[int | str] = {1, "A", 3}
a_list: list[int | str] = [1, "A", 3]
a_dict: dict[str, int | str] = {"hey": "A", "there": 4}
a_tuple: tuple[int, str, int] = (1, "A", 3)
````

Scalar types are pretty easy to use and understand. When you need to store a string value, it will be obvious which
variable type to reach for. Actually, Python will just let you store the value. No need to reach for anything! Knowing
which container type to reach for is a bit harder. When you've been writing Python for a while, you'll get a "feel" for
it. Until then, make mistakes! If you use a dictionary instead of a list, that mistake will become clear at some point
in your development cycle. Refactor the code to use the right type. It will cement the lesson in your mind.

A few starter thoughts that should work until you develop a feel for it:

- Sets
    - Unordered.
    - Automatically removes duplicates.
    - The fastest container type.
    - There's no easy way to reference a specific element in a set by value or position (as sets are unordered)
- Lists
    - Ordered.
    - Faster than dictionaries, slower than sets.
    - Lists can be referenced by index.
- Dictionaries
    - Ordered in more recent Python versions. Prior to Python 3.10 or something, order wasn't maintained.
    - Slower than both sets and lists.
    - If you need to reference the value later by value, this is a great way option.
- Tuples
    - Immutable. Once you create one, you can't alter a tuple. This makes a tuple a tricky "container" as you would need
      to know all the potential values prior to creating it. It's also hard to reference items by value. What it is good
      for is better served by other types we'll cover in later chapters. I don't reach for tuples often.

# Control Logic

Control logic lets us write more dynamic code. Here are a few examples:

````python
# If

mood = "Other"

if mood == "Good":
    print("I'm good!")
elif mood == "Bad":
    print("I'm not so good.")
else:
    print("I'm complicated!")
````

````python
days: list[str] = ["Monday", "Tuesday", "Wednesday", "Thursday", "Friday", "Saturday", "Sunday"]

for day in days:
    print(day)
````

````python
counter: int = 0

while counter < 10:
    print(counter)
    counter += 1
````

````python
# Match statements were introduced in Python 3.10
day = "Monday"

match day:
    case "Friday":
        print("It's Friday!")
    case "Thursday":
        print("It's almost Friday!")
    case _:
        print("Friday is still a ways off")
````

# Functions

Functions are excellent! They have a few benefits:

- **Functions make your code easier to read.** Repeating the same logic in multiple places increases the length of your
  code and the mental energy required to understand it. Once you understand a function, you can reuse that understanding
  when you come across it again.
- **Functions make your code easier to maintain.** Imagine you write identical code in multiple places across your code
  base. That's likely fine today, but future features or maintenance might not be applied to all the copies. Now, you'll
  have __similar__ code in a few places. That's less than helpful, as __similar__ code can appear identical on first
  glance.
- **Functions make your code easier to debug.** If you've already seen the same 12 - 15 lines of code, you're more
  likely to skim and miss subtle differences. When you refactor that code into a function, those subtle differences will
  be `if / else` statements that will make the differences explicit.
- **Functions can be used to narrate your code.** Functions and function parameters require names. These names can serve
  as documentation. If you see `write_results_to_csv(results=some_result, csv_file="my_filename.csv")`, you know what's
  happening without even seeing the function definition.

We've already been using a function: `print`. This is a built-in function in Python that prints to your console. We
don't know how it works, nor do we need to. We can just call the function. Let's define a function:

````python
def my_first_function():
    print("Welcome to my first function!")


my_first_function()
````

## Function: Parameters

Let's add a parameter. Functions can take any number of parameters. When passing a parameter to a function, you can
either pass it as a positional argument or a keyword argument, or `kwarg`:

````python
def my_second_function(name: str):
    print(f"Welcome to my first function, {name}!")


# positional argument
my_second_function("Mike")
# keyword argument
my_second_function(name="Bob")
````

If you try to pass parameters to a function that doesn't take them, Python will throw an exception:

````python
def my_third_function():
    print(f"Welcome to my first function!")


# This will fail
my_third_function("Mike")
````

You can use the special `*args` or `**kwargs` parameter to capture all of the parameters that were passed to the function and not explicitly named:

````python
def my_fourth_function(name: str, *args, **kwargs):
    print(f"Welcome to my first function, {name} - {args} {kwargs}!")


my_fourth_function("Mike", "extra", "extra2", 1, 2, 3, something="else")
````

Note that we can't pick up the `name` argument as a positional and a keyword argument:

````python
def my_fifth_function(name: str, *args, **kwargs):
    print(f"Welcome to my first function, {name} - {args} {kwargs}!")


my_fifth_function("Mike", "extra", "extra2", 1, 2, 3, name="Mike", something="else")
````

My recommendation: **Use keyword arguments whenever possible.**

Which of these two code snippets would you rather read:

```python
result = forier_transform(img, {"grid_x": 4, "grid_y": 6})
```

```python
result = forier_transform(
    input_image=img,
    transform_parameters={"grid_x": 4, "grid_y": 6}
)
```

The second one, right?

It's not just about readability. If you're calling a function from a 3rd-party library, it's possible that the order the
positional arguments changes from one version to another. That would be called out in the release documentation, but
we're not always referencing that documentation when we upgrade packages. Using keyword arguments insulates you from
this. Mature 3rd-party libraries rarely make changes like this. On the other hand, the functions that **you** define may
change significantly with each commit. The benefit of keyword arguments is especially pronounced here.

## Function: Return Values

The functions we've defined so far don't return anything. We can make that explicitly by hinting the return type. (Note
the `-> None`)

````python
def my_sixth_function() -> None:
    print("Welcome to my first function!")


my_sixth_function()
````

Your function can return a value:

````python
def addition(a: int, b: int) -> int:
    return a + b


result = addition(1, 2)
print(result)
````

Your function can also return multiple values, but it will need to do so using some container:

````python
def addition_tuple(a: int, b: int) -> tuple[int, int, int]:
    return a, b, a + b


def addition_list(a: int, b: int) -> list[int]:
    return [a, b, a + b]


def addition_dict(a: int, b: int) -> dict[str, int]:
    return {"a": a, "b": b, "result": a + b}


result = addition_tuple(1, 2)
print(result)

result = addition_list(1, 2)
print(result)

result = addition_dict(1, 2)
print(result)
````

# Scope

When we introduced control logic functions, we introduced another shadow concept: scope. Every Python script has a
`global` scope. A gross simplification: Every indent introduces a new nested scope. Every dedent returns us back to the
parent scope.

The `global` scope can either be modified when you're in the global scope (no indentation) or by using the `global`
keyword to let Python know that you're referring to a global variable. **Altering global variables isn't considered a
best practice, so we won't be doing it in this program.** I only mention it because you might come across this behavior
as you are exploring the language.

````python
x = 1

print(f"global x = {x}")


def some_function():
    # This variable name is "shadowing" another variable name. Your IDE might complain about this.
    x = 2
    y = 2
    print(f"some_function: local x = {x}")
    print(f"some_function: local y = {y}")


some_function()
# The value of x is unchanged. We didn't tell Python that we wanted to change a global variable
print(f"global x = {x}")
# Your IDE should complain about this. This might fail on some operating systems. It's considered "undefined behavior" and should be avoided
print(
    f"global y = {y}")


def another_function():
    global x
    x = 3
    print(f"another_function: local x = {x}")


another_function()
# The value of x changed.
print(f"global x = {x}")
````

Scopes are pretty simple. I contrived this example to show you some odd behavior. Generally, you'll stay out of trouble
if you avoid referencing a variable defined in a child scope from a parent scope. Also, leave global variables as
constants if you use them at all.

# By Value / By Reference

When you pass an argument into a function in Python, are you passing the original value or a copy of the original value?
First, an example:

````python
is_workday_lookup: dict[str, bool] = {"monday": True, "tuesday": True, "wednesday": True, "thursday": True}


def is_today_a_workday(today: str, lookup: dict[str, bool]):
    is_workday = lookup.pop(today)
    if is_workday:
        print(f"today is a workday")
    else:
        print(f"today is not a workday")


is_today_a_workday(today="monday", lookup=is_workday_lookup)
````

This function works, right? Let's look a bit closer:

````python
is_workday_lookup: dict[str, bool] = {"monday": True, "tuesday": True, "wednesday": True, "thursday": True}


def is_today_a_workday(today: str, lookup: dict[str, bool]):
    is_workday = lookup.pop(today)
    if is_workday:
        print(f"today is a workday")
    else:
        print(f"today is not a workday")


print(f"before: {is_workday_lookup}")
is_today_a_workday(today="monday", lookup=is_workday_lookup)
print(f"after: {is_workday_lookup}")
````

When we pop from the lookup parameter, we're actually popping from the underlying lookup! That's likely not the desired
behavior. Since the Battlesnake `move` payload is a dictionary, you need to be careful not to alter it or delete
anything from it. If you need to do something destructive to the payload, you can use the `deepcopy.copy()` library to
create an __actual__ copy instead of a reference to the original.
